```python
%%writefile app.py
import streamlit as st
import pandas as pd
import plotly.express as px

st.set_page_config(page_title="Zomato Dashboard", layout="wide")

st.title("🍽️ Zomato Data Storytelling Dashboard")

uploaded_file = st.file_uploader("Upload Zomato CSV", type=["csv"])

if uploaded_file:
    df = pd.read_csv(uploaded_file, encoding="latin1")

    df.drop_duplicates(inplace=True)
    df.dropna(inplace=True)

    df.columns = df.columns.str.lower().str.replace(" ", "_")

    if 'rate' in df.columns:
        df['rate'] = df['rate'].astype(str).str.replace('/5', '')
        df['rate'] = pd.to_numeric(df['rate'], errors='coerce')

    if 'approx_cost(for_two_people)' in df.columns:
        df['approx_cost(for_two_people)'] = pd.to_numeric(
            df['approx_cost(for_two_people)'], errors='coerce'
        )

    st.subheader("📊 Dataset Preview")
    st.dataframe(df.head())

    st.subheader("📌 KPIs")
    col1, col2, col3 = st.columns(3)

    col1.metric("Total Restaurants", len(df))

    if 'rate' in df.columns:
        col2.metric("Avg Rating", round(df['rate'].mean(), 2))

    if 'approx_cost(for_two_people)' in df.columns:
        col3.metric("Avg Cost", round(df['approx_cost(for_two_people)'].mean(), 2))

    st.divider()

    if 'rest_type' in df.columns:
        st.subheader("🍽️ Top Restaurant Types")
        top = df['rest_type'].value_counts().head(10).reset_index()
        top.columns = ['rest_type', 'count']
        fig = px.bar(top, x='rest_type', y='count')
        st.plotly_chart(fig, use_container_width=True)

    if 'rate' in df.columns:
        st.subheader("⭐ Rating Distribution")
        fig = px.histogram(df, x='rate', nbins=20)
        st.plotly_chart(fig, use_container_width=True)

    if 'approx_cost(for_two_people)' in df.columns:
        st.subheader("💰 Cost Distribution")
        fig = px.histogram(df, x='approx_cost(for_two_people)', nbins=30)
        st.plotly_chart(fig, use_container_width=True)

    if 'location' in df.columns:
        st.subheader("🏙️ Top Locations")
        top_city = df['location'].value_counts().head(10).reset_index()
        top_city.columns = ['city', 'count']
        fig = px.bar(top_city, x='city', y='count')
        st.plotly_chart(fig, use_container_width=True)

    if 'online_order' in df.columns:
        st.subheader("🛵 Online Orders")
        online = df['online_order'].value_counts().reset_index()
        online.columns = ['online_order', 'count']
        fig = px.pie(online, names='online_order', values='count')
        st.plotly_chart(fig)

    if 'book_table' in df.columns:
        st.subheader("📅 Table Booking")
        book = df['book_table'].value_counts().reset_index()
        book.columns = ['book_table', 'count']
        fig = px.pie(book, names='book_table', values='count')
        st.plotly_chart(fig)

    st.subheader("💡 Insights")
    st.write("""
    - Most restaurants are in major cities
    - Mid-range restaurants dominate
    - Online ordering is widely available
    - Ratings are mostly between 3 and 4.5
    - Table booking is less common
    """)

    st.subheader("📌 Conclusion")
    st.write("""
    - Improve low-rated restaurants
    - Expand in new locations
    - Promote online ordering
    - Focus on affordable restaurants
    - Improve customer satisfaction
    """)
```
